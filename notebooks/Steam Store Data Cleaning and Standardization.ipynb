# **CLEANING THE DATASET**
**Pricing**

```python
# Clean original_price
df['original_price_clean'] = (
    df['original_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .replace('nan', np.nan)
    .replace('', np.nan)
)
df['original_price_clean'] = pd.to_numeric(df['original_price_clean'], errors='coerce')
```

```python
# Clean discounted_price (handle "Free" games)
df['discounted_price_clean'] = (
    df['discounted_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .replace('Free', '0')
    .replace('nan', np.nan)
    .replace('', np.nan)
)
df['discounted_price_clean'] = pd.to_numeric(df['discounted_price_clean'], errors='coerce')
```

```python
# Clean discount_percentage (negative values mean discount)
df['discount_pct_clean'] = pd.to_numeric(
    df['discount_percentage'].astype(str).str.replace('%', ''),
    errors='coerce'
)
```

```python
# Convert to absolute value (so -50% becomes 50%)
df['discount_pct_clean'] = df['discount_pct_clean'].abs()
```

```python
# Create flags
df['is_free_to_play'] = df['discounted_price_clean'] == 0
df['is_on_sale'] = df['discount_pct_clean'] > 0
```

```python
# Final price (use discounted if available, else original)
df['final_price'] = df['discounted_price_clean'].fillna(df['original_price_clean'])
```

```python
print("Price cleaning summary:")
print(f"Free games: {df['is_free_to_play'].sum()}")
print(f"Games on sale: {df['is_on_sale'].sum()}")
print(f"Games with pricing: {df['final_price'].notna().sum()}")
```

**Dates and compute age**

```python
# Parse release_date
df['release_date_clean'] = pd.to_datetime(df['release_date'], format='%d %b, %Y', errors='coerce')
df['release_year'] = df['release_date_clean'].dt.year
df['release_month'] = df['release_date_clean'].dt.month
```

```python
# Game age (as of end of 2024)
snapshot = datetime(2024, 12, 31)
df['game_age_years'] = ((snapshot - df['release_date_clean']).dt.days / 365.25)
```

```python
# Age buckets for analysis
df['age_bucket'] = pd.cut(
    df['game_age_years'],
    bins=[0, 1, 3, 5, 10, 100],
    labels=['<1y', '1-3y', '3-5y', '5-10y', '10y+'],
    include_lowest=True
)
```

```python
print(f"Date parsing: {df['release_date_clean'].notna().sum()} valid dates")
```

**Clean genres and categories**

```python
# Standardize genres
df['genres_clean'] = (
    df['genres']
    .astype(str)
    .str.lower()
    .str.replace(', ', ';')
    .replace('nan', np.nan)
)
```

```python
# Extract main genre (first one)
df['main_genre'] = (
    df['genres_clean']
    .str.split(';')
    .str[0]
    .str.strip()
)
```

```python
# Same for categories
df['categories_clean'] = (
    df['categories']
    .astype(str)
    .str.lower()
    .replace('nan', np.nan)
)
```

```python
print(f"Unique main genres: {df['main_genre'].nunique()}")
print("\nTop 10 genres:")
print(df['main_genre'].value_counts().head(10))
```

```python
df
```

**Clean review metrics**

```python
# Rename for clarity
df = df.rename(columns={
    'overall_review_%': 'overall_review_pct',
    'recent_review_%': 'recent_review_pct'
})
```

```python
# Ensure numeric
for col in ['overall_review_pct', 'overall_review_count',
            'recent_review_pct', 'recent_review_count']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# Log transform for review counts (reduce skew)
df['log_overall_review_count'] = np.log1p(df['overall_review_count'].fillna(0))
df['log_recent_review_count'] = np.log1p(df['recent_review_count'].fillna(0))
```

```python
print("Review data summary:")
print(f"Games with overall reviews: {df['overall_review_count'].notna().sum()}")
print(f"Games with recent reviews: {df['recent_review_count'].notna().sum()}")
```

```python
df
```

**Create business-relevant flags**

```python
# Platform support flags already boolean
# Create multi-platform flag
df['multi_platform'] = (
    df['win_support'].astype(int) +
    df['mac_support'].astype(int) +
    df['linux_support'].astype(int)
) > 1
```

```python
# DLC flag
df['has_dlc'] = df['dlc_available'] > 0
```

```python
# Early Access games (from genres)
df['is_early_access'] = df['genres_clean'].fillna('').str.contains('early access')
```

```python
# Age rating categories
df['age_rating_category'] = pd.cut(
    df['age_rating'],
    bins=[-1, 0, 6, 12, 16, 100],
    labels=['All Ages', '7+', '13+', '17+', '18+']
)
```

```python
print("\nBusiness flags:")
print(f"Multi-platform games: {df['multi_platform'].sum()}")
print(f"Games with DLC: {df['has_dlc'].sum()}")
print(f"Early Access: {df['is_early_access'].sum()}")
```

**Missing data review**

```python
# Check missingness after cleaning
missing = df[[
    'final_price', 'discount_pct_clean', 'release_year', 'game_age_years',
    'main_genre', 'overall_review_count', 'recent_review_count'
]].isna().mean().sort_values(ascending=False)

print("Missingness after cleaning:")
print(missing)
```

# EDA Visualizations
**Univariate distributions**

```python
sns.set(style="whitegrid", palette="muted")
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# Price distribution (exclude free games)
paid_games = df[df['final_price'] > 0]
axes[0,0].hist(paid_games['final_price'], bins=50, edgecolor='black')
axes[0,0].set_title('Final Price Distribution (Paid Games)')
axes[0,0].set_xlabel('Price (₹)')
axes[0,0].set_ylabel('Count')

# Log review count
axes[0,1].hist(df['log_overall_review_count'].dropna(), bins=50, edgecolor='black')
axes[0,1].set_title('Log Overall Review Count')
axes[0,1].set_xlabel('log1p(review_count)')

# Release year
df['release_year'].value_counts().sort_index().plot(kind='bar', ax=axes[0,2])
axes[0,2].set_title('Games by Release Year')
axes[0,2].set_xlabel('Year')
axes[0,2].tick_params(axis='x', rotation=90)

# Review percentage
axes[1,0].hist(df['overall_review_pct'].dropna(), bins=50, edgecolor='black')
axes[1,0].set_title('Overall Review % (Positive Ratio)')
axes[1,0].set_xlabel('Positive %')

# Game age
axes[1,1].hist(df['game_age_years'].dropna(), bins=50, edgecolor='black')
axes[1,1].set_title('Game Age Distribution')
axes[1,1].set_xlabel('Age (years)')

# DLC count
axes[1,2].hist(df['dlc_available'], bins=30, edgecolor='black')
axes[1,2].set_title('DLC Count Distribution')
axes[1,2].set_xlabel('Number of DLCs')

plt.tight_layout()
plt.show()
```

**Top genres (business question: which types dominate?)**

```python
top_genres = df['main_genre'].value_counts().head(15)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_genres.values, y=top_genres.index, palette='viridis')
plt.title('Top 15 Genres by Number of Games', fontsize=14, fontweight='bold')
plt.xlabel('Number of Games')
plt.ylabel('Genre')
plt.tight_layout()
plt.show()
```

**Price vs engagement (business question: revenue vs popularity)**

```python
# Only games with price and reviews
analysis_df = df[(df['final_price'] > 0) & (df['overall_review_count'] > 0)].copy()

plt.figure(figsize=(10, 6))
plt.scatter(
    analysis_df['final_price'],
    analysis_df['log_overall_review_count'],
    alpha=0.3,
    s=20
)
plt.title('Price vs Engagement (Log Reviews)', fontsize=14, fontweight='bold')
plt.xlabel('Final Price (₹)')
plt.ylabel('log1p(Overall Review Count)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Engagement by genre (business question: which genres generate most engagement?)**

```python
genre_engagement = (
    df.groupby('main_genre')['overall_review_count']
    .agg(['median', 'count'])
    .sort_values('median', ascending=False)
    .head(15)
    .reset_index()
)

fig, ax1 = plt.subplots(figsize=(12, 7))

ax1.barh(genre_engagement['main_genre'], genre_engagement['median'], color='steelblue')
ax1.set_xlabel('Median Overall Review Count', fontsize=12)
ax1.set_ylabel('Genre', fontsize=12)
ax1.set_title('Top 15 Genres by Median Engagement', fontsize=14, fontweight='bold')
ax1.invert_yaxis()

plt.tight_layout()
plt.show()
```

**Older titles performance (business question: do older games still generate engagement?)**

```python
age_engagement = (
    df.groupby('age_bucket')['overall_review_count']
    .agg(['median', 'mean', 'count'])
    .reset_index()
)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(age_engagement['age_bucket'].astype(str), age_engagement['median'], color='coral')
axes[0].set_title('Median Review Count by Game Age', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Age Bucket')
axes[0].set_ylabel('Median Review Count')

axes[1].bar(age_engagement['age_bucket'].astype(str), age_engagement['count'], color='teal')
axes[1].set_title('Number of Games by Age', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Age Bucket')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

**Discount impact (business question: how do discounts relate to engagement?)**

```python
discount_comparison = df[df['final_price'] > 0].groupby('is_on_sale').agg({
    'overall_review_count': 'median',
    'overall_review_pct': 'median',
    'app_id': 'count'
}).reset_index()
discount_comparison.columns = ['is_on_sale', 'median_reviews', 'median_positive_pct', 'count']

print("Discount vs No Discount:")
print(discount_comparison)

plt.figure(figsize=(10, 5))
sns.boxplot(data=df[df['final_price'] > 0], x='is_on_sale', y='log_overall_review_count', palette='Set2')
plt.title('Engagement: Discounted vs Full Price Games', fontsize=14, fontweight='bold')
plt.xlabel('On Sale')
plt.ylabel('log1p(Overall Review Count)')
plt.tight_layout()
plt.show()
```
